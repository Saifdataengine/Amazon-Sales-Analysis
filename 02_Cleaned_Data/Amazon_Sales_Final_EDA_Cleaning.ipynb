```python
# ==========================================
# AMAZON SALES DATA - FINAL CLEANING
# ==========================================

import pandas as pd
import numpy as np


# ==========================================
# 1. Load Initial Cleaned Dataset
# ==========================================

df = pd.read_csv('amazon_sales_cleaned.csv')

print("Initial Shape:", df.shape)


# ==========================================
# 2. Drop Unnecessary Column
# ==========================================

if "Unnamed: 22" in df.columns:
    df.drop(columns=["Unnamed: 22"], inplace=True)

if "index" in df.columns:
    df.drop(columns=["index"], inplace=True)


# ==========================================
# 3. Date Cleaning
# ==========================================

df["Date"] = pd.to_datetime(df["Date"], errors="coerce")


# ==========================================
# 4. Postal Code Cleaning
# ==========================================

df["ship-postal-code"] = df["ship-postal-code"].astype(str)

df["ship-postal-code"] = df["ship-postal-code"].str.replace(
    ".0", "", regex=False
)


# ==========================================
# 5. Currency Missing Values
# ==========================================

df["currency"] = df["currency"].fillna("INR")


# ==========================================
# 6. Promotion IDs Missing Values
# ==========================================

df["promotion-ids"] = df["promotion-ids"].fillna("No Promotion")


# ==========================================
# 7. Fulfilled By Missing Values
# ==========================================

df["fulfilled-by"] = df["fulfilled-by"].fillna("Unknown")


# ==========================================
# 8. Remove Rows Where Complete Shipping
#    Location is Missing
# ==========================================

df = df.dropna(
    subset=[
        "ship-city",
        "ship-state",
        "ship-postal-code",
        "ship-country"
    ]
)


# ==========================================
# 9. Date Based Columns
# ==========================================

df["Year"] = df["Date"].dt.year

df["Month"] = df["Date"].dt.month

df["Month_Name"] = df["Date"].dt.month_name()

df["Day"] = df["Date"].dt.day

df["Day_Name"] = df["Date"].dt.day_name()


# ==========================================
# 10. Promotion Status
# ==========================================

df["Promotion_Status"] = df["promotion-ids"].apply(
    lambda x: "No Promotion"
    if x == "No Promotion"
    else "Promotion"
)


# ==========================================
# 11. Final Missing Value Check
# ==========================================

print("\nMissing Values:")
print(df.isna().sum())


# ==========================================
# 12. Final Dataset Information
# ==========================================

print("\nFinal Shape:", df.shape)

print("\nData Types:")
print(df.dtypes)

```
